## 4. Data Quality Assessment

Check missing values and duplicate rows before deciding whether any cleaning is required.

### 4.1 Check missing values

```python
missing_count = df.isnull().sum()
print("Missing values by column:")
print(missing_count)
print("\nTotal missing values:", missing_count.sum())
```

### 4.2 Check duplicate records

## 5. Target Variable Analysis

The target variable is `G3`, the student's final mathematics grade. This is a regression task because the target is a numerical grade.

### 5.1 Distribution of final grades (`G3`)

```python
grade_distribution = df["G3"].value_counts().sort_index()
print("Final grade frequency:")
print(grade_distribution)
```

### 5.2 Summary statistics for final grades

```python
print("Mean final grade:", round(df["G3"].mean(), 2))
print("Median final grade:", df["G3"].median())
print("Minimum final grade:", df["G3"].min())
print("Maximum final grade:", df["G3"].max())
```
