```python
import os
import sys
import joblib
import pandas as pd
import numpy as np

from pathlib import Path

print("DAY 11 - PROJECT FINALIZATION AND QA")
print("=" * 55)

# Project directories
PROJECT_ROOT = Path("..").resolve()
DATASET_DIR = PROJECT_ROOT / "dataset"
NOTEBOOK_DIR = PROJECT_ROOT / "notebooks"
ARTIFACT_DIR = PROJECT_ROOT / "artifacts"

print("Python Version:", sys.version)
print("Project Root:", PROJECT_ROOT)
print("Dataset Directory:", DATASET_DIR)
print("Notebook Directory:", NOTEBOOK_DIR)
print("Artifacts Directory:", ARTIFACT_DIR)

print("\nDirectory Verification")
print("=" * 40)

for name, path in [
    ("Project Root", PROJECT_ROOT),
    ("Dataset", DATASET_DIR),
    ("Notebooks", NOTEBOOK_DIR),
    ("Artifacts", ARTIFACT_DIR)
]:
    print(f"{name}: {'Available' if path.exists() else 'Missing'}")
```

```python
# Inspect project directory contents

print("PROJECT FILE STRUCTURE AUDIT")
print("=" * 60)

directories = {
    "DATASET": DATASET_DIR,
    "NOTEBOOKS": NOTEBOOK_DIR,
    "ARTIFACTS": ARTIFACT_DIR
}

for section, directory in directories.items():
    print(f"\n{section}")
    print("-" * 60)

    files = sorted(directory.iterdir())

    if not files:
        print("No files found.")
        continue

    for file in files:
        file_type = "DIR " if file.is_dir() else "FILE"
        size = f"{file.stat().st_size:,} bytes" if file.is_file() else ""
        print(f"[{file_type}] {file.name} {size}")

print("\nPROJECT ROOT FILES")
print("-" * 60)

for file in sorted(PROJECT_ROOT.iterdir()):
    if file.is_file():
        print(f"{file.name} ({file.stat().st_size:,} bytes)")
```

```python
# Verify critical project files

required_files = {
    "Dataset": DATASET_DIR / "student-mat.csv",
    "Student Information": DATASET_DIR / "student.txt",
    "Main Streamlit App": PROJECT_ROOT / "app.py",
    "Optimized Pipeline": ARTIFACT_DIR / "day07_optimized_pipeline.pkl",
    "Day 10 Metrics": ARTIFACT_DIR / "day10_evaluation_metrics.csv",
    "Day 10 Predictions": ARTIFACT_DIR / "day10_predictions.csv",
    "Day 11 Notebook": NOTEBOOK_DIR / "10_Project_Finalization_and_QA.ipynb"
}

print("CRITICAL FILE VERIFICATION")
print("=" * 60)

for name, path in required_files.items():
    status = "AVAILABLE" if path.exists() else "MISSING"
    print(f"{name}: {status}")

# Inspect notebook sizes
print("\nNOTEBOOK SIZE CHECK")
print("=" * 60)

for notebook in sorted(NOTEBOOK_DIR.glob("*.ipynb")):
    size_kb = notebook.stat().st_size / 1024
    print(f"{notebook.name}: {size_kb:.2f} KB")

# Verify app.py contents
print("\nSTREAMLIT APPLICATION CHECK")
print("=" * 60)

app_path = PROJECT_ROOT / "app.py"
app_content = app_path.read_text(encoding="utf-8")

for section in [
    "Family Background",
    "Academic Information",
    "Lifestyle and Well-being",
    "Prediction History",
    "download_button"
]:
    print(f"{section}: {'Found' if section in app_content else 'Not Found'}")
```

```python
# Load and verify dataset
dataset = pd.read_csv(DATASET_DIR / "student-mat.csv", sep=";")

# Load optimized pipeline
model_path = ARTIFACT_DIR / "day07_optimized_pipeline.pkl"
optimized_pipeline = joblib.load(model_path)

print("DATASET VALIDATION")
print("=" * 50)
print("Dataset shape:", dataset.shape)
print("Missing values:", dataset.isnull().sum().sum())
print("Duplicate records:", dataset.duplicated().sum())
print("Target column available:", "G3" in dataset.columns)

print("\nOPTIMIZED PIPELINE VALIDATION")
print("=" * 50)
print("Pipeline type:", type(optimized_pipeline).__name__)
print("Pipeline steps:", list(optimized_pipeline.named_steps.keys()))

# Prepare beginning-of-term features
excluded_columns = ["G1", "G2", "G3", "absences"]
X = dataset.drop(columns=excluded_columns)
y = dataset["G3"]

# Test prediction on a small sample
sample_input = X.head(5)
sample_predictions = optimized_pipeline.predict(sample_input)

print("\nPREDICTION WORKFLOW TEST")
print("=" * 50)
print("Input shape:", sample_input.shape)
print("Prediction count:", len(sample_predictions))
print("Predictions:", np.round(sample_predictions, 2))
print("Actual grades:", y.head(5).tolist())
print("All predictions finite:", np.isfinite(sample_predictions).all())
```
